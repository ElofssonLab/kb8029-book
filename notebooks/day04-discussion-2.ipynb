## Why does affine care about scattering, but linear doesn't?

Consider a single gap of total length 4, placed two different ways:
as **one contiguous block** of 4, or as **four separate single-residue
gaps** scattered elsewhere in the alignment.

**As a group, using $g(n) = n \times g_{\text{op}}$ for linear and
$g(n) = g_{\text{op}} + (n-1) \times g_{\text{ext}}$ for affine:**

1. Linear, with $g_{\text{op}} = -1$: what's the total cost of the 4-block? Of the four scattered single gaps?
2. Affine, with $g_{\text{op}} = -8$, $g_{\text{ext}} = -1$: what's the total cost of the 4-block? Of the four scattered single gaps? (Each scattered gap is its own isolated 1-residue gap, so each one pays its own opening cost.)
3. Which scheme is indifferent between the two placements? Which one clearly prefers keeping the gap as one block?

Run the cell below to check your arithmetic.

```python
def linear_penalty(n, g_op):
    return n * g_op

def affine_penalty(n, g_op, g_ext):
    return g_op + (n - 1) * g_ext

g_op_lin = -1
g_op_aff, g_ext_aff = -8, -1

print("Linear scheme (g_op = -1):")
print("  one 4-block:          ", linear_penalty(4, g_op_lin))
print("  four scattered 1-gaps:", 4 * linear_penalty(1, g_op_lin))

print()
print("Affine scheme (g_op = -8, g_ext = -1):")
print("  one 4-block:          ", affine_penalty(4, g_op_aff, g_ext_aff))
print("  four scattered 1-gaps:", 4 * affine_penalty(1, g_op_aff, g_ext_aff))
```

**Discussion point:** linear charges -4 either way — it genuinely
doesn't care how the gap is distributed, because its cost is purely
proportional to total gap length. Affine charges -11 for the
consolidated block but -32 for the scattered version — nearly 3&times;
worse — because each isolated gap has to pay its own $g_{\text{op}}$
all over again. This is why affine produces biologically
plausible alignments where indels stay consolidated into blocks, the
way real insertions/deletions actually happen.

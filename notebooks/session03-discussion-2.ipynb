## How long is human haemoglobin beta?

A classmate wants the length of human haemoglobin beta (HBB, 147 residues). Their script searches
UniProtKB for `gene:HBB AND organism_id:9606` and prints the mean length of the hits. What does it
print?

**Fix and run.** The cell below runs the search on a copy of the results saved on 14 October 2026
(the last cell runs it live). Run it once as it is, and look at the table: which entries came back,
and who vouches for them? Then change the line marked `FIX` so that only the entries reviewed by
curators remain, and run it again.

````python
import io, os
import pandas as pd
import requests

URL = "https://raw.githubusercontent.com/ElofssonLab/kb8029-book/main/notebooks/data/session03-hbb-uniprot.tsv"
LOCAL = "data/session03-hbb-uniprot.tsv"
text = open(LOCAL).read() if os.path.exists(LOCAL) else requests.get(URL, timeout=30).text
table = pd.read_csv(io.StringIO(text), sep="\t")   # UniProtKB: gene:HBB AND organism_id:9606

hits = table                                        # FIX: keep only the reviewed entries
print(hits[["Entry", "Reviewed", "Length", "Protein names"]].to_string(index=False, max_colwidth=45))
print(f"\n{len(hits)} hits, mean length {hits['Length'].mean():.1f}")
````

````python
# The same search, live from UniProt (needs the internet; the result may have grown)
r = requests.get("https://rest.uniprot.org/uniprotkb/search",
                 params={"query": "gene:HBB AND organism_id:9606", "fields": "accession,reviewed,length",
                         "format": "tsv", "size": 100}, timeout=30)
live = pd.read_csv(io.StringIO(r.text), sep="\t")
print(f"{len(live)} hits today, {sum(live['Reviewed'] == 'reviewed')} reviewed")
````

**Discussion point.** As it was, the script printed 18 hits with a mean length of 80.4: one
reviewed entry (Swiss-Prot `P68871`, `HBB_HUMAN`, 147 residues) and 17 unreviewed TrEMBL entries,
most of them fragments and variants submitted from individual studies (18 to 128 residues). With

```python
hits = table[table["Reviewed"] == "reviewed"]
```

only `P68871` remains, and the mean is 147.

The search was not wrong: every hit really is annotated as human HBB. But a reviewed entry is one a
curator has checked against the literature, one per protein, while unreviewed entries are added
automatically, one per submitted sequence. The 147-residue TrEMBL entries are the same protein filed
again, which UniParc would show as one sequence. When you want *the* protein, ask for the reviewed
entry of the right organism, and check what came back before you compute anything with it.

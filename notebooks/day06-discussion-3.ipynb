## Is "x1 AND NOT x2" linearly separable?

Consider a new function of two binary inputs: it outputs 1 only when
$x_1 = 1$ **and** $x_2 = 0$, and 0 otherwise.

| $x_1$ | $x_2$ | output |
|---|---|---|
| 0 | 0 | 0 |
| 0 | 1 | 0 |
| 1 | 0 | 1 |
| 1 | 1 | 0 |

**As a group:**

1. Using the same style of reasoning as the book's XOR proof (or just
   geometrically), is this function linearly separable?
2. If so, find weights $w_1, w_2, b$ that make a perceptron
   ($z = w_1 x_1 + w_2 x_2 + b > 0 \Rightarrow 1$, else 0) implement it.
   If not, show why not.
3. Discuss: what's geometrically different about this function compared
   to XOR, that makes the answer come out differently?

Run the cell below to check your answer.

```python
w1, w2, b = 1.0, -1.0, -0.5   # <-- try your own group's answer here first

truth = {(0, 0): 0, (0, 1): 0, (1, 0): 1, (1, 1): 0}

for (x1, x2), exp in truth.items():
    z = w1 * x1 + w2 * x2 + b
    out = 1 if z > 0 else 0
    print(f"x=({x1},{x2})  z={z:>5.1f}  output={out}  expected={exp}  {'OK' if out == exp else 'WRONG'}")
```

**Discussion point:** $w_1=1, w_2=-1, b=-0.5$ works — this function
*is* linearly separable, unlike XOR. Geometrically, only one corner of
the input square, $(1,0)$, is positive; a single straight line can
easily cut that one corner off from the other three. XOR fails
precisely because its two "positive" corners sit *diagonally opposite*
each other, with the two "negative" corners on the other diagonal — no
straight line can separate a diagonal pair from the other diagonal
pair, no matter where you draw it. "x1 AND NOT x2" has no such
diagonal conflict, which is why it's solvable and XOR isn't.

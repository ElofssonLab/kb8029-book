## Predict the Boolean count before running it

Two new search terms this time — not the hemoglobin/malaria pair from the book page. Without running anything yet:

1. Predict what `CRISPR OR cancer` should return, given `CRISPR` alone, `cancer` alone, and `CRISPR AND cancer`, using the same inclusion-exclusion idea as the book page.
2. Predict `CRISPR NOT cancer`.
3. Would `cancer NOT CRISPR` give the *same* number as `CRISPR NOT cancer`? Work out why or why not before checking.

```python
import requests

def pubmed_count(term):
    r = requests.get(
        "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/esearch.fcgi",
        params={"db": "pubmed", "term": term, "retmode": "json"},
        timeout=10,
    )
    r.raise_for_status()
    return int(r.json()["esearchresult"]["count"])

a = pubmed_count("CRISPR")
b = pubmed_count("cancer")
a_and_b = pubmed_count("CRISPR AND cancer")
a_or_b = pubmed_count("CRISPR OR cancer")
a_not_b = pubmed_count("CRISPR NOT cancer")
b_not_a = pubmed_count("cancer NOT CRISPR")

print(f"CRISPR                {a:>10,}")
print(f"cancer                {b:>10,}")
print(f"CRISPR AND cancer     {a_and_b:>10,}")
print(f"CRISPR OR cancer      {a_or_b:>10,}   (predicted: {a + b - a_and_b:,})")
print(f"CRISPR NOT cancer     {a_not_b:>10,}   (predicted: {a - a_and_b:,})")
print(f"cancer NOT CRISPR     {b_not_a:>10,}   (predicted: {b - a_and_b:,})")
```

**Discussion point:** `OR` and both `NOT` directions check out exactly against set-theory arithmetic. But `CRISPR NOT cancer` and `cancer NOT CRISPR` are **very different numbers** — `NOT` is not symmetric, because it always starts from one whole set (say, everything about CRISPR) and only removes the overlap, leaving a result the size of *that* set minus the shared part. Since `CRISPR` and `cancer` individually are wildly different sizes, subtracting the same small overlap from each leaves very different remainders.

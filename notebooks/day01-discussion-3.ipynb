## Segment 3: How much should you trust a plagiarism-detection flag?

A university's automated system flags 85% of truly plagiarized
assignments ($P(\text{flag}\mid\text{plagiarized}) = 0.85$), and
incorrectly flags 8% of genuine assignments
($P(\text{flag}\mid\text{not plagiarized}) = 0.08$). Historically,
about 3% of submitted assignments are actually plagiarized.

**As a group:**

1. Write down what each number above means, and what you're being asked to find, in $P(\cdot\mid\cdot)$ notation. Which direction of conditioning does the student care about?
2. Count it out: of 10,000 assignments, how many are plagiarized, how many of those are flagged, and how many genuine assignments are flagged? What fraction of the flagged assignments is plagiarized, $P(\text{plagiarized}\mid\text{flag})$?
3. Would you recommend automatically failing every flagged student based on this number alone? Why or why not?

Run the cell below to check your answer to (2). (If you read the optional Bayes box on the Day 1 page: the same number comes out of Bayes' rule.)

```python
n = 10_000
plagiarized = 0.03 * n
genuine = n - plagiarized
flagged_plag = 0.85 * plagiarized
flagged_genuine = 0.08 * genuine

print(f"plagiarized: {plagiarized:.0f}, flagged among them: {flagged_plag:.0f}")
print(f"genuine: {genuine:.0f}, flagged among them: {flagged_genuine:.0f}")
p_plag_given_flag = flagged_plag / (flagged_plag + flagged_genuine)
print(f"P(plagiarized | flag) = {flagged_plag:.0f} / {flagged_plag + flagged_genuine:.0f} = {p_plag_given_flag:.3f} = {p_plag_given_flag*100:.1f}%")
```

**Discussion point:** despite a detector that catches 85% of cases, a
flagged assignment is plagiarized only about 25% of the time. Plagiarism
is rare, so the false positives from the large honest majority (776)
outnumber the true positives from the small dishonest minority (255).
$P(\text{flag}\mid\text{plagiarized})$ and $P(\text{plagiarized}\mid\text{flag})$
are very different numbers: the direction of conditioning matters.

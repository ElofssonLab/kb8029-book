## Which vector points most in the direction of x?

Take $\mathbf{x} = (2, 1)$ and three vectors:

- $\mathbf{a} = (3, 4)$, with $\mathbf{a} \cdot \mathbf{x} = 10$
- $\mathbf{b} = (30, 40)$, with $\mathbf{b} \cdot \mathbf{x} = 100$
- $\mathbf{c} = (4, 2)$, with $\mathbf{c} \cdot \mathbf{x} = 10$

Which of them points in the most similar direction to $\mathbf{x}$, that is, makes the smallest angle
with it?

**Fix and run.** The cell below is meant to print $\cos\theta$ for each vector: 1 means the same
direction, 0 means a right angle. The line marked `FIX` computes something else. Run it once as it
is: which vector would you pick from this output? Then fix the line and run it again.

````python
import numpy as np

x = np.array([2., 1.])
vectors = {"a": np.array([3., 4.]), "b": np.array([30., 40.]), "c": np.array([4., 2.])}

for name, w in vectors.items():
    cos = w @ x                  # FIX: this is not yet cos θ
    print(f"{name}: cos θ = {cos:.3f}")
````

As it was, the cell printed the dot product itself, and $\mathbf{b}$
looked best by far (100 against 10). The fix divides by both lengths:

```python
cos = w @ x / (np.linalg.norm(w) * np.linalg.norm(x))
```

Then $\mathbf{c}$ gives $\cos\theta = 1.000$ (it points exactly along $\mathbf{x}$), while
$\mathbf{a}$ and $\mathbf{b}$ both give 0.894 (27° away). $\mathbf{b}$ is just $\mathbf{a}$ ten times
longer, so its dot product is ten times larger at the same angle; $\mathbf{a}$ and $\mathbf{c}$ have
the same dot product but different directions. The dot product mixes length and direction,
$\mathbf{w}\cdot\mathbf{x} = \lVert\mathbf{w}\rVert\lVert\mathbf{x}\rVert\cos\theta$; dividing by the
lengths (normalising) leaves only the angle.

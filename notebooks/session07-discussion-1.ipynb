## What happens when two training examples disagree?

A perceptron with one input and no bias ($z = wx$), squared-error loss, starting at $w_0 = 0$,
learning rate $\eta = 0.02$, but now **two** training examples at once, $(x, y) = (2, 6)$ and
$(x, y) = (3, 3)$, with total loss $L(w) = (2w - 6)^2 + (3w - 3)^2$.

**As a group:**

1. Is there any single value of $w$ that makes *both* predictions exact ($2w = 6$ and $3w = 3$
   simultaneously)?
2. Compute $dL/dw$ at $w = 0$ by hand, and take one gradient descent step.
3. Predict: where does $w$ eventually converge, if no value fits both examples perfectly?

**Fix and run.** The cell below runs gradient descent, but one character in the line marked `FIX`
is wrong: run it once as it is and look at what happens to $w$. Then fix it and run it again.

```python
xs, ys = [2, 3], [6, 3]          # the two examples (x, y)
w, eta = 0.0, 0.02

def grad(w):
    # dL/dw for L(w) = (2w - 6)^2 + (3w - 3)^2
    return sum(2 * (w*x - y) * x for x, y in zip(xs, ys))

print("gradient at w = 0:", grad(0.0))
for step in range(1000):
    w = w + eta * grad(w)        # FIX: one character in this line is wrong
    if step < 5:
        print(f"step {step + 1}: w = {w:.3f}")
print("after 1000 steps: w =", round(w, 2))
```

**Discussion point.** The mistake was the sign: gradient descent steps *against* the gradient,
`w = w - eta * grad(w)`. With `+` it climbs the loss, and $w$ runs away, further with every step.

No single weight satisfies both examples: $2w = 6$ needs $w = 3$, but $3w = 3$ needs $w = 1$. The
gradient at $w = 0$ is $-42$, so the first step goes to $w = 0.84$. Gradient descent does not get
stuck or fail here: it converges to $w^* \approx 1.62$, the value that minimises the *total* squared
error. That is not the plain average 2. The loss is $4(w - 3)^2 + 9(w - 1)^2$, and the second
example, with the larger $x$, is steeper, so it pulls harder:
$w^* = (4 \cdot 3 + 9 \cdot 1) / 13 = 21/13$.

This is the situation real data sets are in all the time: not every example can be fitted, and the
model settles on the best balance its capacity allows (here one weight, no bias).

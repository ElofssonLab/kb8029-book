## Does calling .backward() twice give you 2x the gradient, or something else?

```python
x = torch.tensor(3.0, requires_grad=True)
y = x**2
y.backward()
print(x.grad)          # first call

y2 = x**2
y2.backward()
print(x.grad)          # second call -- note: no zero_grad() in between
```

**Fix and run.** Run the cell below once as it is: what does the second `print` show? Then add one
line at the place marked `FIX`, so that the second `backward()` gives 6 again, and run it again.

**If you finish early.** By hand: what is dy/dx at x = 3 for y = x³? Change the code to check it.

````python
import torch

x = torch.tensor(3.0, requires_grad=True)
y = x**2
y.backward()
print("grad after 1st backward():", x.grad.item())

# FIX: add one line here, so that the second backward() starts from zero
y2 = x**2
y2.backward()
print("grad after 2nd backward():", x.grad.item())
````

<details>
<summary><b>Discussion point</b> &mdash; open this once you have run the code and talked it through</summary>

dy/dx = 2x = 6 at x = 3, so the first
`backward()` correctly sets `x.grad` to 6. But PyTorch **accumulates**
gradients into `.grad` by default rather than replacing them — the
second `backward()` computes another 6 and *adds* it to what's already
there, giving 12, not 6. This is why every training loop you'll
see from here on calls something like `optimizer.zero_grad()` (or
`w.grad.zero_()`, as Session 7's notebook does directly) at the start of
every iteration — without it, gradients from previous steps silently
keep piling up. The fix is `x.grad.zero_()` before the second
`backward()` (or `x.grad = None`). For y = x³, dy/dx = 3x² = 27 at x = 3.

<summary><b>Discussion point</b> &mdash; open this once you have run the code and talked it through</summary>

The mistake is on **line 3**, but Python reports the error on
**line 4**:

```
----> 4     lengths = [len(r.seq) for r in SeqIO.parse(handle, "fasta")]
...
StreamModeError: Fasta files must be opened in text mode.
```

`gzip.open(path)` opens the file in binary mode (`"rb"`) by default, so the handle gives raw bytes.
Nothing goes wrong until `SeqIO.parse` tries to read text from it, on line 4. The fix is on line 3:

```python
with gzip.open("globins.fasta.gz", "rt") as handle:     # "rt": read as text
```

With the fix, the code prints 147.67: haemoglobin beta has 147 residues, alpha 142 and myoglobin 154.

**The general lesson.** The line in an error message is where Python *noticed* the problem, not
necessarily where you made it. Read the last line of the traceback first (what went wrong), then ask
where the object it complains about came from: here, the handle made on line 3.

</details>

````python
# The fixed version
from Bio import SeqIO
import gzip
with gzip.open("globins.fasta.gz", "rt") as handle:
    lengths = [len(r.seq) for r in SeqIO.parse(handle, "fasta")]
print(sum(lengths) / len(lengths))
````

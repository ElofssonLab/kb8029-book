## How much should you trust a plagiarism flag?

Every submitted assignment is checked by an automatic plagiarism detector:

- it flags 85% of the assignments that are plagiarised, $P(\text{flag}\mid\text{plagiarised}) = 0.85$;
- it also wrongly flags 8% of the genuine ones, $P(\text{flag}\mid\text{genuine}) = 0.08$;
- 3% of all assignments are plagiarised.

Your assignment is flagged. How likely is it that it really is plagiarised,
$P(\text{plagiarised}\mid\text{flag})$?

**Fix and run.** The cell counts 10 000 assignments, but the line marked `FIX` divides by the wrong
number. Run it once as it is: which probability does it print? Then fix it and run it again.

````python
n = 10_000
plagiarised = 0.03 * n
genuine = n - plagiarised
flagged_plagiarised = 0.85 * plagiarised
flagged_genuine = 0.08 * genuine
print(f"plagiarised: {plagiarised:.0f}, flagged among them: {flagged_plagiarised:.0f}")
print(f"genuine: {genuine:.0f}, flagged among them: {flagged_genuine:.0f}")

p = flagged_plagiarised / plagiarised    # FIX: divide by the right group
print(f"P(plagiarised | flag) = {p:.3f}")
````

As it was, the cell divided by the plagiarised assignments and printed
0.850: that is $P(\text{flag}\mid\text{plagiarised})$, the wrong direction of conditioning. The flag
restricts us to the *flagged* assignments:

```python
p = flagged_plagiarised / (flagged_plagiarised + flagged_genuine)
```

which gives $255 / (255 + 776) = 0.247$, about 25%. Plagiarism is rare, so the false positives from
the large honest majority (776) outnumber the true positives from the small dishonest minority
(255). $P(\text{flag}\mid\text{plagiarised})$ and $P(\text{plagiarised}\mid\text{flag})$ are very
different numbers: the direction of conditioning matters. With 30% plagiarised, a flag would mean
82%; for the rare disease, a positive test means only 9%.

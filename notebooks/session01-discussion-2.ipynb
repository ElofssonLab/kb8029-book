## What happens with a learning rate that is too large?

The loss is $L(w) = (w - 4)^2$, with its minimum at $w = 4$. We start at $w = 0$. With learning rate
$\eta = 0.5$, one gradient descent step lands exactly on the minimum. What happens with $\eta = 2$ if
we keep stepping?

**Fix and run.** The cell below runs five steps with each learning rate, but the derivative in the
line marked `FIX` is wrong. Run it once as it is: does $\eta = 0.5$ land on 4 in one step? Then fix
the derivative and run it again.

```python
def grad(w):
    return w - 4                 # FIX: the derivative of (w - 4)**2 is not w - 4

for eta in (0.5, 2.0):
    w = 0.0
    path = [w]
    for step in range(5):
        w = w - eta * grad(w)
        path.append(w)
    print(f"eta = {eta}:", "  ".join(f"{v:g}" for v in path))
```

**If you finish early.**

1. Write the update as $w_{n+1} = w_n - 2\eta(w_n - 4)$. Which value of $\eta$ makes the factor in
   front of $(w_n - 4)$ exactly zero after the step?
2. For which learning rates does gradient descent still converge, perhaps zig-zagging? Try
   $\eta = 0.9$ and $\eta = 1.1$.

**Discussion point.** The derivative of $(w - 4)^2$ is $2(w - 4)$: the chain rule gives the
factor 2. Without it, $\eta = 0.5$ only gets halfway at each step (2, 3, 3.5, ...), and $\eta = 2$
jumps between 0 and 8 for ever: a tempting wrong answer.

With the right derivative, the update is $w_{n+1} - 4 = (1 - 2\eta)(w_n - 4)$. For $\eta = 0.5$ the
factor is 0, so one step lands on 4. For $\eta = 2$ it is $-3$: each step overshoots three times as
far, on alternating sides (0, 16, −32, 112, ...), and $w$ moves away from the minimum. Any
$\eta$ between 0.5 and 1 zig-zags but converges, since the factor is between −1 and 0. This is the
textbook picture of a learning rate that is too large.

## Does a hidden layer help without an activation function?

A tiny network with two hidden units and a ReLU activation, $\text{ReLU}(t) = \max(0, t)$:

$$
h_1 = \text{ReLU}(x_1 + x_2), \qquad h_2 = \text{ReLU}(x_1 + x_2 - 1), \qquad z = h_1 - 2 h_2
$$

(in matrix form: $W_1 = \begin{pmatrix} 1 & 1 \\ 1 & 1 \end{pmatrix}$, $\mathbf{b}_1 = (0, -1)$,
$W_2 = (1, -2)$, $b_2 = 0$).

**As a group:**

1. Compute $z$ by hand for the four inputs $(0,0), (0,1), (1,0), (1,1)$. Which logic function is it?
2. Now remove the ReLU: $h_1 = x_1 + x_2$ and $h_2 = x_1 + x_2 - 1$. Write $z$ as a *single* weighted
   sum $w_1 x_1 + w_2 x_2 + b$, and compute it for the four inputs. Is it still the same function?
3. Could *any* choice of weights and biases, with no activation between the layers, compute XOR?
   Why, or why not?

**Fix and run.** In the cell below, the line marked `FIX` does not compute a ReLU: run it once as it
is (that is question 2), then fix it and run it again (question 1).

```python
import numpy as np

W1 = np.array([[1., 1.], [1., 1.]]); b1 = np.array([0., -1.])
W2 = np.array([1., -2.]);            b2 = 0.0
relu = lambda t: t                   # FIX: this should be ReLU(t) = max(0, t)

print(" x       z    XOR")
for x in [(0, 0), (0, 1), (1, 0), (1, 1)]:
    x = np.array(x, dtype=float)
    z = W2 @ relu(W1 @ x + b1) + b2
    print(f"{x.astype(int)}  {z:5.1f}    {int(x[0]) ^ int(x[1])}")
```

```python
# Without an activation, two layers are one: W2 (W1 x + b1) + b2 = (W2 W1) x + (W2 b1 + b2)
w, b = W2 @ W1, W2 @ b1 + b2
print("without ReLU, the network is the single weighted sum z =", w, "· x +", b)

# The same holds for any weights: check with random ones
rng = np.random.default_rng(0)
A1, c1, A2, c2 = rng.normal(size=(2, 2)), rng.normal(size=2), rng.normal(size=2), rng.normal()
x = rng.normal(size=2)
print("random weights, two linear layers vs one:", np.isclose(A2 @ (A1 @ x + c1) + c2, (A2 @ A1) @ x + (A2 @ c1 + c2)))
```

**Discussion point:** with the ReLU (`relu = lambda t: np.maximum(0, t)`), the network computes
XOR exactly: $z = 0, 1, 1, 0$. Without it, the two layers collapse into one weighted sum,
$z = -x_1 - x_2 + 2$, which gives $2, 1, 1, 0$: not XOR. That is not bad luck with these particular
weights. For *any* weights, $W_2(W_1 \mathbf{x} + \mathbf{b}_1) + b_2 = (W_2 W_1)\mathbf{x} + (W_2
\mathbf{b}_1 + b_2)$ is a single linear function of the inputs, and the page proves that no linear
function can compute XOR. The nonlinear activation between the layers is what makes a hidden layer
worth having: here it "switches off" $h_2$ for every input except $(1,1)$, and that switch is exactly
the interaction XOR needs.

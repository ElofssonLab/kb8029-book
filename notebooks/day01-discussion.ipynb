## Segment 1: Does a bigger dot product mean a better match?

Two teams each report a "match score" between an observed vector
$\mathbf{x} = (2, 1)$ and their own profile's weight vector, using a
plain dot product:

- Team A's profile: $\mathbf{w}_A = (3, 4)$
- Team B's profile: $\mathbf{w}_B = (30, 40)$

**As a group:**

1. Compute $\mathbf{w}_A \cdot \mathbf{x}$ and $\mathbf{w}_B \cdot \mathbf{x}$ by hand.
2. Based only on those two numbers, does Team B's profile match $\mathbf{x}$ better?
3. Compute the length (norm) of each weight vector, then normalize each one to a unit vector (length 1) and recompute the dot product with $\mathbf{x}$ using the *normalized* weights.
4. What do you notice? What does this tell you about comparing dot-product "scores" computed from vectors of different magnitude?

Run the cell below to check your answers.

```python
import numpy as np

w_A = np.array([3., 4.])
w_B = np.array([30., 40.])
x = np.array([2., 1.])

print("raw dot product, A:", w_A @ x)
print("raw dot product, B:", w_B @ x)

u_A = w_A / np.linalg.norm(w_A)
u_B = w_B / np.linalg.norm(w_B)
print("\n||w_A|| =", np.linalg.norm(w_A), " unit vector:", u_A)
print("||w_B|| =", np.linalg.norm(w_B), " unit vector:", u_B)

print("\nnormalized dot product, A:", u_A @ x)
print("normalized dot product, B:", u_B @ x)
```

**Discussion point:** $\mathbf{w}_B$ is just $\mathbf{w}_A$ scaled up
by 10 — same direction, same underlying "profile," just written with
bigger numbers. The raw dot product is fooled by that scaling; the
normalized (unit-vector) dot product isn't, because dividing by the
length removes exactly the $\lVert\mathbf{w}\rVert$ factor from
$\mathbf{w}\cdot\mathbf{x} = \lVert\mathbf{w}\rVert\lVert\mathbf{x}\rVert\cos\theta$,
leaving only the angle — which is genuinely the same for both.

## Segment 2: A surprisingly good (and a surprisingly bad) learning rate

A 2-parameter model has loss $L(w_1, w_2) = (w_1 - 4)^2 + (w_2 + 2)^2$,
currently at $\mathbf{w} = (0, 0)$.

**As a group:**

1. Compute $\nabla L$ at $(0, 0)$ by hand.
2. Using learning rate $\eta = 0.5$, compute one gradient descent step to get $\mathbf{w}'$.
3. Without further computation: what do you predict happens if you kept repeating this update?
4. Bonus: what happens instead if $\eta = 2$?

Run the cell below to check your answers.

```python
import numpy as np

def grad(w1, w2):
    return np.array([2 * (w1 - 4), 2 * (w2 + 2)])

w = np.array([0., 0.])
g = grad(*w)
print("gradient at (0,0):", g)

eta = 0.5
w_next = w - eta * g
print("after one step (eta=0.5):", w_next, " <- exactly the minimum, (4, -2)!")

print()
w = np.array([0., 0.])
eta2 = 2.0
for n in range(5):
    print(f"n={n}  w={w}")
    w = w - eta2 * grad(*w)
```

**Discussion point:** with $\eta = 0.5$, one step lands *exactly* on
the minimum $(4, -2)$ — no surprise once you see why: the update becomes
$w_{n+1} = w_n - 2\eta(w_n - a)$, and $2\eta = 1$ exactly cancels the
distance to $a$. With $\eta = 2$, the same formula gives a coefficient
of $-3$, so each step overshoots the minimum by 3&times; and flips sign
— an oscillation whose magnitude grows without bound. This is the
textbook picture of "learning rate too large."

## Segment 3: How much should you trust a plagiarism-detection flag?

A university's automated system flags 85% of truly plagiarized
assignments ($P(\text{flag}\mid\text{plagiarized}) = 0.85$), and
incorrectly flags 8% of genuine assignments
($P(\text{flag}\mid\text{not plagiarized}) = 0.08$). Historically,
about 3% of submitted assignments are actually plagiarized.

**As a group:**

1. Write down what each number above means, and what you're being asked to find, in $P(\cdot\mid\cdot)$ notation.
2. Apply Bayes' rule to compute $P(\text{plagiarized}\mid\text{flag})$.
3. Would you recommend automatically failing every flagged student based on this number alone? Why or why not?

Run the cell below to check your answer to (2).

```python
p_plag = 0.03
p_flag_given_plag = 0.85
p_flag_given_noplag = 0.08

p_flag = p_flag_given_plag * p_plag + p_flag_given_noplag * (1 - p_plag)
p_plag_given_flag = p_flag_given_plag * p_plag / p_flag

print("P(flag):", round(p_flag, 4))
print("P(plagiarized | flag):", round(p_plag_given_flag, 4), f"= {p_plag_given_flag*100:.1f}%")
```

**Discussion point:** despite an 85%-accurate detector, a flagged
assignment is actually plagiarized only about 25% of the time — because
plagiarism is rare, false positives from the large honest majority
outnumber true positives from the small dishonest minority. The same
arithmetic as Day 1's disease-test example, with real stakes attached.
